```python
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np

def get_path(symb: str) -> str:
	return f"./data/{symb}_data.parquet"

qqq_etf: str = "QQQ"
spy_etf: str = "SPY"

qqq_data: pd.DataFrame = pd.read_parquet(get_path(qqq_etf))
spy_data: pd.DataFrame = pd.read_parquet(get_path(spy_etf))
```

```python
qqq_spy_plot: go.Figure = go.Figure()
qqq_spy_plot.add_trace(
	go.Scatter(x = qqq_data.index, y = qqq_data.vwap, name = "QQQ")
)
qqq_spy_plot.add_trace(
	go.Scatter(x = spy_data.index, y = spy_data.vwap, name = "SPY")
)
qqq_spy_plot.show()
```

```python
difference_plot: go.Figure = go.Figure()
difference_plot.add_trace(go.Scatter(x = qqq_data.index, y = spy_data.vwap - qqq_data.vwap))
```

```python
ratio_plot: go.Figure = go.Figure(go.Scatter(x = qqq_data.index, y = spy_data.vwap / qqq_data.vwap))
ratio_plot.show()
```

```python
xy_plot: go.Figure = go.Figure(go.Scatter(x = spy_data.vwap, y = qqq_data.vwap))
xy_plot.show()
print(spy_data.vwap.corr(qqq_data.vwap))
```
